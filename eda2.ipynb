# Product Prices by Active Status

This notebook uses the cleaned `silver/products.csv` file. The figure compares unit price and cost price for active and inactive products.

```python
import pandas as pd
import matplotlib.pyplot as plt

products = pd.read_csv('silver/products.csv')
```

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

products.boxplot(column='unit_price', by='is_active', ax=axes[0])
axes[0].set_title('Unit price by active status')
axes[0].set_xlabel('Active status')
axes[0].set_ylabel('Unit price')

products.boxplot(column='cost_price', by='is_active', ax=axes[1])
axes[1].set_title('Cost price by active status')
axes[1].set_xlabel('Active status')
axes[1].set_ylabel('Cost price')

fig.suptitle('Product prices by active status')
plt.tight_layout()
plt.show()
```
